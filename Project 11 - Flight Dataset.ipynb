# 1. What are the airlines in the dataset, accompanied by their frequencies ?

```python
data.head(2)
```

```python
data.airline.unique()
```

```python
# SHowing all the airlines with their number of flights in the bar graph
```

```python
data['airline'].value_counts().sort_values(ascending = True).plot.bar(color = ['pink', 'yellow', 'lightgreen', 'lightblue', 'brown', 'purple'])
plt.title("Airlines with frequencies")
plt.xlabel("Number of flights")
plt.ylabel("Airline")

plt.show()
```

# 2. Show bar graph representing the departure time & arrival time ?

```python
data.head(2)
```

```python
data['departure_time'].value_counts()
```

```python
data['arrival_time'].value_counts()
```

```python
plt.figure(figsize = (16,4))
plt.subplot(1,2,1)

#for departure_time
plt.bar(data['departure_time'].value_counts().index, data['departure_time'].value_counts().values, color = ['pink', 'yellow', 'lightgreen', 'lightblue', 'brown', 'purple'])

plt.title("Departure Time")
plt.xlabel("D. Time")
plt.ylabel("D. Freq")

plt.show()
```

```python
#for arrival_time
plt.figure(figsize = (16,4))
plt.subplot(1,2,2)

plt.bar(data['arrival_time'].value_counts().index, data['arrival_time'].value_counts().values, color = ['pink', 'yellow', 'lightgreen', 'lightblue', 'brown', 'purple'])

plt.title("Arrival Time")
plt.xlabel("D. Time")
plt.ylabel("D. Freq")

plt.show()
```
