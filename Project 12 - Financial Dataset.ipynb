```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
%matplotlib inline
```

# 1. How much amount a company spent on R & D?

```python
data.head()
```

```python
print("Company's spending for R & D in $Bn")
RD = data.groupby('Company')['R&D_Spending_USD_Mn'].sum()/1000
RD
```

```python
# show in a bar plot
plt.bar(RD.index, RD.values, color = ['pink', 'yellow', 'orange'])

plt.title("R&D Spending by the companies")
plt.xlabel("Company")
plt.ylabel("Amount in USD_$bn")

plt.show()
```

# 2. Revenue earned by the companies ?

```python
data.head(2)
```

```python
print("Revenue earned by the Company")
rev = data.groupby('Company')['AI_Revenue_USD_Mn'].sum()/1000
rev
```

```python
# show in a bar plot
plt.bar(rev.index, rev.values, color = ['pink', 'yellow', 'orange'], width = 0.3)

plt.title("Revenue earned by the Company")
plt.xlabel("Company")
plt.ylabel("Revenue Earned")

plt.show()
```

```python
# bar plot to show expenditure on r&d and revenue of the companies.

plt.figure(figsize = (10,4))

plt.subplot(1,2,1)

plt.bar(RD.index, RD.values, color = ['pink', 'yellow', 'orange'])

plt.title("R&D Spending by the companies")
plt.xlabel("Company")
plt.ylabel("Amount Spent in USD_$bn")

plt.subplot(1,2,2)

plt.bar(rev.index, rev.values, color = ['pink', 'yellow', 'orange'], width = 0.3)

plt.title("Revenue earned by the Company")
plt.xlabel("Company")
plt.ylabel("Revenue Earned in USD_$bn")

plt.show()
```

# 9. Expenditure vs Revenue year-by-year

```python
data.head()
```

```python
spend = data.groupby('Year')['R&D_Spending_USD_Mn'].sum()
spend
```

```python
# showing the amount spent on R&D
plt.plot(spend.index, spend.values, color = 'green')

plt.title("Amount spent on R&D")
plt.xlabel("Year")
plt.ylabel("Amount")

plt.show()
```

```python
rev_earned = data.groupby('Year')['AI_Revenue_USD_Mn'].sum()
rev_earned
```

```python
# showing the amount spent on R&D
plt.plot(rev_earned.index, rev_earned.values, color = 'red')

plt.title("Revenue earned growth")
plt.xlabel("Year")
plt.ylabel("Amount")

plt.show()
```

```python
# combined graph

plt.plot(spend.index, spend.values, color = 'green')
plt.plot(rev_earned.index, rev_earned.values, color = 'red')

plt.title("Combined Expenditure vs Revenue year-by-year", fontsize = 10)
plt.xlabel("Year")
plt.ylabel("Amount")

plt.legend(['Expenditure', 'Revenue'])

plt.show
```
