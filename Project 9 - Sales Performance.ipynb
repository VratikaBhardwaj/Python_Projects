# 2. Most selling product? - By Quantity and By Revenue

```python
data.head(2)
```

```python
# By Quantity
data.groupby('Product')['Quantity'].sum().sort_values(ascending = False)
```

# Very Important
- The given below is how to convert the result into columns to form a dataframe. Basically to convert the index of a series into a column to form a dataframe

```python
most_quantity = data.groupby('Product')['Quantity'].sum().sort_values(ascending = False)
```

```python
most_quantity
```

```python
type(most_quantity)
```

```python
most_quantity = most_quantity.reset_index()
most_quantity
```

```python
plt.figure(figsize = (9,4))
plt.bar(most_quantity['Product'], most_quantity['Quantity'], color = ['pink', 'yellow', 'orange', 'cyan', 'grey'])
plt.title("Most Selling Product By Quantity")
plt.xlabel("Product")
plt.ylabel("Quantity")
```

```python
# By Revenue
data['Revenue'] = data['Quantity']*data['Price']
```

```python
data
```

```python
data.groupby('Product').Revenue.sum().sort_values(ascending = False)
```

```python
most_revenued_product = data.groupby('Product').Revenue.sum().sort_values(ascending = False)
```

```python
most_revenued_product
```

```python
type(most_revenued_product)
```

```python
most_revenued_product = most_revenued_product.reset_index()
most_revenued_product
```

```python
plt.figure(figsize = (9,4))
plt.bar(most_revenued_product['Product'], most_revenued_product['Revenue'], color = ['pink', 'yellow', 'orange', 'cyan', 'grey'])
plt.title("Most Selling Product By Revenue")
plt.xlabel("Product")
plt.ylabel("Revenue")
```
