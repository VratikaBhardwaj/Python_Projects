# 15. Identify departments with the highest attrition rate (Resigned %).

```python
data.head(2)
```

```python
data.groupby('Department').agg( total_emp=('Status', 'count'), resigned=('Status', lambda x: (x == 'Resigned').sum()))
```

```python
dept_counts = data.groupby('Department').agg( total_emp=('Status', 'count'), resigned=('Status', lambda x: (x == 'Resigned').sum()))
```

```python
dept_counts
```

```python
# calculate attrition rate
dept_counts['attrition_rate%'] = (dept_counts['resigned']/dept_counts['total_emp']) * 100
```

```python
dept_counts
```

```python
#sort by highest attrition rate
dept_counts.sort_values("attrition_rate%", ascending = False)
```
